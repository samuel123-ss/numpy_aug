```python
#create a 4x4 matrix with float values with z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])

z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(np.floor(z))
print(np.ceil(z))
print(np.round(z))
print(z.max())
print(z.min())
print(z.sum())

```

```python

z=np.array([[3.1,4.5,7.7,3.5],[2.9,6.3,5.5,8.5],[1.6,2.7,3.7,8.2]])
print(z.max(axis=1))
print(z.max(axis=0))
print(np.sqrt(z))
print(np.sin(z))
print(np.tan(z))
print(np.cos(z))
```
